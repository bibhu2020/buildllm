# NLP Evolution Timeline

| Era                            | Years (Approx.)     | Approach                                   | Key Limitation                           |
| ------------------------------ | ------------------- | ------------------------------------------ | ---------------------------------------- |
| Rule-based NLP                 | 1950s – early 1990s | Hand-written linguistic rules              | No scalability, brittle                  |
| Statistical NLP                | Early 1990s – ~2010 | Probabilistic models (n-grams, HMMs, CRFs) | Short context, heavy feature engineering |
| Word Embeddings + Classical ML | 2008 – 2014         | Word2Vec, GloVe + SVM/CRF                  | No contextual meaning                    |
| RNN-based Models               | 2010 – 2014         | RNN, LSTM, GRU                             | Poor long-range memory                   |
| Seq2Seq (Encoder–Decoder)      | 2014 – 2015         | RNN/LSTM encoder–decoder                   | Single-vector bottleneck                 |
| Attention Mechanisms           | 2015 – 2017         | Cross-attention in Seq2Seq                 | Sequential, slow training                |
| Transformers                   | 2017 – present      | Self-attention–based models                | Compute and data intensive               |
